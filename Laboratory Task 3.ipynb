````python
import numpy as np

# input and target
x = np.array([1, 0, 1])
y = 1

# learning rate
lr = 0.001

# hidden weights
w11 = 0.2
w12 = -0.3
w13 = 0.4
w14 = 0.1
w15 = -0.5
w16 = 0.2

# output weights
w21 = -0.3
w22 = -0.2

# biases
theta1 = -0.4
theta2 = 0.2
theta3 = 0.1


# ReLU
def relu(z):
    return max(0, z)


# ReLU derivative
def relu_derivative(z):
    if z > 0:
        return 1
    else:
        return 0


# -----------------
# FORWARD PASS
# -----------------

z1 = w11*x[0] + w13*x[1] + w15*x[2] + theta1
h1 = relu(z1)

z2 = w12*x[0] + w14*x[1] + w16*x[2] + theta2
h2 = relu(z2)

z3 = w21*h1 + w22*h2 + theta3
yhat = relu(z3)

# output error
error = yhat - y

print("FORWARD PASS")
print("z1 =", z1)
print("h1 =", h1)
print("z2 =", z2)
print("h2 =", h2)
print("z3 =", z3)
print("prediction =", yhat)
print("error =", error)


# -----------------
# BACKWARD PASS
# -----------------

# output delta
delta_output = error * relu_derivative(z3)

# gradients for output weights
dw21 = delta_output * h1
dw22 = delta_output * h2

# gradient for output bias
dtheta3 = delta_output

# hidden deltas
delta_h1 = delta_output * w21 * relu_derivative(z1)
delta_h2 = delta_output * w22 * relu_derivative(z2)

# hidden weight gradients
dw11 = delta_h1 * x[0]
dw13 = delta_h1 * x[1]
dw15 = delta_h1 * x[2]

dw12 = delta_h2 * x[0]
dw14 = delta_h2 * x[1]
dw16 = delta_h2 * x[2]

# hidden bias gradients
dtheta1 = delta_h1
dtheta2 = delta_h2


# -----------------
# UPDATE WEIGHTS
# -----------------

w21 = w21 - lr * dw21
w22 = w22 - lr * dw22
theta3 = theta3 - lr * dtheta3

w11 = w11 - lr * dw11
w13 = w13 - lr * dw13
w15 = w15 - lr * dw15

w12 = w12 - lr * dw12
w14 = w14 - lr * dw14
w16 = w16 - lr * dw16

theta1 = theta1 - lr * dtheta1
theta2 = theta2 - lr * dtheta2


print("\nBACKWARD PASS")
print("Output delta =", delta_output)

print("\nUpdated weights:")
print("w21 =", w21)
print("w22 =", w22)

print("w11 =", w11)
print("w12 =", w12)
print("w13 =", w13)
print("w14 =", w14)
print("w15 =", w15)
print("w16 =", w16)

print("\nUpdated biases:")
print("theta1 =", theta1)
print("theta2 =", theta2)
print("theta3 =", theta3)
````
