$$x = \begin{bmatrix} 1 \\ 0 \\ 1 \end{bmatrix}$$

$$y = \begin{bmatrix} 1 \end{bmatrix}$$

$$f=max(0, Z_n)$$

$$\text{hidden unit weights} = 
\begin{bmatrix}
w_{11} = 0.2 && w_{12} = -0.3 \\
w_{13} = 0.4 && w_{14} = 0.1 \\
w_{15} = -0.5 && w_{16} = 0.2
\end{bmatrix}$$

$$\text{output unit weights} = 
\begin{bmatrix}
w_{21} = -0.3 \\
w_{22} = -0.2
\end{bmatrix}$$

$$\theta = 
\begin{bmatrix}
\theta_{1} = -0.4 \\
\theta_{2} = 0.2 \\
\theta_{3} = 0.1
\end{bmatrix}$$
</div>

```python
import numpy as np

# input
x = np.array([1, 0, 1])
y = 1

# hidden weights
w11 = 0.2
w12 = -0.3
w13 = 0.4
w14 = 0.1
w15 = -0.5
w16 = 0.2

# output weights
w21 = -0.3
w22 = -0.2

# biases
theta1 = -0.4
theta2 = 0.2
theta3 = 0.1

# ReLU function
def relu(z):
    return max(0, z)

# forward pass
z1 = w11*x[0] + w13*x[1] + w15*x[2] + theta1
h1 = relu(z1)

z2 = w12*x[0] + w14*x[1] + w16*x[2] + theta2
h2 = relu(z2)

z3 = w21*h1 + w22*h2 + theta3
yhat = relu(z3)

# error
error = yhat - y

print("Z1 =", z1)
print("H1 =", h1)
print("Z2 =", z2)
print("H2 =", h2)
print("Z3 =", z3)
print("Prediction =", yhat)
print("Error =", error)
```
