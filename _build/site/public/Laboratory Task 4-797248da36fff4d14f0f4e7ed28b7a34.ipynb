```python
import torch
import torch.nn as nn
import numpy as np

from torch.utils.data import TensorDataset, DataLoader
```

```python
# Input data
x_train = np.array([
    [73, 67, 43],
    [91, 88, 64],
    [87, 134, 58],
    [102, 43, 37],
    [69, 96, 70],
    [73, 67, 43],
    [91, 88, 64],
    [87, 134, 58],
    [102, 43, 37],
    [69, 96, 70],
    [73, 67, 43],
    [91, 88, 64],
    [87, 134, 58],
    [102, 43, 37],
    [69, 96, 70]
], dtype='float32')

# Target data
y_train = np.array([
    [56, 70],
    [81, 101],
    [119, 133],
    [22, 37],
    [103, 119],
    [56, 70],
    [81, 101],
    [119, 133],
    [22, 37],
    [103, 119],
    [56, 70],
    [81, 101],
    [119, 133],
    [22, 37],
    [103, 119]
], dtype='float32')
```

```python
inputs = torch.from_numpy(x_train / 100)
targets = torch.from_numpy(y_train / 100)

print(inputs.shape)
print(targets.shape)
```

```python
train_ds = TensorDataset(inputs, targets)

batch_size = 8

train_dl = DataLoader(
    train_ds,
    batch_size=batch_size,
    shuffle=True
)
```

```python
model = nn.Sequential(
    nn.Linear(3, 8),
    nn.ReLU(),
    nn.Linear(8, 2)
)

print(model)
```

```python
criterion = nn.MSELoss()

optimizer = torch.optim.SGD(
    model.parameters(),
    lr=0.01
)
```

```python
epochs = 1000

for epoch in range(epochs):

    for xb, yb in train_dl:

        # prediction
        prediction = model(xb)

        # calculate loss
        loss = criterion(prediction, yb)

        # remove old gradients
        optimizer.zero_grad()

        # calculate new gradients
        loss.backward()

        # update weights
        optimizer.step()

    if (epoch + 1) % 100 == 0:
        print(
            "Epoch:", epoch + 1,
            "Loss:", loss.item()
        )
```

```python
with torch.no_grad():

    prediction = model(inputs)

    print("Predicted values:")
    print(prediction * 100)

    print("\nActual values:")
    print(targets * 100)
```

# Conclusion


The model was able to learn from the given data using PyTorch. With two fully connected layers, MSE Loss, SGD optimizer, a batch size of 8, and 1000 epochs, the training process gradually reduced the loss and made the predicted values closer to the actual values. This shows how a simple neural network can be trained to make predictions from a regression dataset.
