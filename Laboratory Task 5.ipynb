```python
import torch
import numpy as np
```

```python
def set_seed(seed: int):
    np.random.seed(seed)
    torch.manual_seed(seed)
```

```python
set_seed(42)

arr = np.random.randint(0, 5, 6).astype('int32')

print(arr)
print(arr.dtype)
```

```python
x = torch.from_numpy(arr)

print(x)
```

```python
x = x.type(torch.int64)

print(x)
print(x.dtype)
```

```python
x = x.reshape(3, 2)

print(x)
```

```python
right_column = x[:, 1]

print(right_column)
```

```python
square_x = x ** 2

print(square_x)
```

```python
torch.manual_seed(42)

y = torch.randint(0, 5, (2, 3))

print(y)
```

```python
result = torch.matmul(x, y)

print(result)
```
